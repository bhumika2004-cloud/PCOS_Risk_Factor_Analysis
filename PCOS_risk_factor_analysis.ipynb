```python
df_clean = df.copy()
```

```python
df_clean = df_clean.drop(columns=["Unnamed: 44"])
```

Removed the unnamed column as it has no valuable data that can be used for analysis.Removed it because it had no column name, only 2 filled cells (a . and a 7), and no way to tell what they meant.

```python
col = "II    beta-HCG(mIU/mL)"
df_clean[col] = df_clean[col].replace("1.99.", "1.99")
df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce")
```

changed the 1.99. to 1.99 and converted the whole column to numbers.Changed 1.99. to 1.99 because the extra period was an obvious typo, so keeping the value was better than losing it."

```python
df_clean["AMH(ng/mL)"] = pd.to_numeric(df_clean["AMH(ng/mL)"], errors="coerce")
```

turned the a into missing value and changed the column to numbers.Turned the a into a missing value because I can't know the real number. Deleting the patient would lose 43 good measurements, and filling in the average would invent data
