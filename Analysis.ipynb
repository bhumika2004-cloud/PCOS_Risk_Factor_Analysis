```python
import pandas as pd
df_clean = pd.read_csv("pcos_clean.csv")
df_clean.shape
```

```python
df_clean['PCOS (Y/N)'].value_counts()
```

With this I counted the number of patients without PCOS and with PCOS.

```python
df_clean.groupby('PCOS (Y/N)')["BMI"].mean()
```

This gave me the mean of BMI of patients without PCOS and mean of BMI of patients with PCOS and on average the mean of BMI of patients with PCOS was more than the mean of BMI of patients without PCOS.

```python
df_clean.groupby('PCOS (Y/N)')["Weight (Kg)"].mean()
```

```python
df_clean.groupby("PCOS (Y/N)")[["Age (yrs)", "BMI", "Weight (Kg)", "AMH(ng/mL)"]].mean().round(2)
```
